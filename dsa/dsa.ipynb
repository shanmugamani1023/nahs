```python
a=[2, 7, 11, 15]
t=9
def find_indexes_of_target_sum(a,t):
    seen={}
    for i,j in enumerate(a):
        complement=t-j
        if complement in seen:
            return i,seen[complement]
        seen[j]=i
res=find_indexes_of_target_sum(a,t)
print(res)
```

```python
# Today's problem — #2: Best Time to Buy and Sell Stock

prices = [7, 1, 5, 3, 6, 4]
max_profit=0
buy_price=0
profit_day=0
stock_value=0
for i,j in enumerate(prices):
    if i==0:
        buy_price=j
        buy_index=i
    if i>0:
        if buy_price > j:
            buy_price=j
            buy_index=i+1
            
        profit=j-buy_price
    
        if profit > max_profit:
            max_profit=profit
            profit_day=i+1
            stock_value=j
    

print(buy_price,stock_value,profit_day,max_profit)
    
    
```

```python
a='[()]'
#valid parenthesis
st=[]
d={')':"(","]":"["}

def _check_(a,d):
    for i in a:
        if i in d:
            if st :
                l=st.pop()
                if l!=d[i]:
                    return False
        else:
            st.append(i)
    return len(st)==0

result=_check_(a,d)
print(result)


    
```

```python
prices = [7, 1, 5, 3, 6, 4]

# Find Best Time to Buy and Sell Stock


# We have to find Track low price , and its index and Best profit and its index

low_price=prices[0]
low_price_index=0

max_profit=0
max_profit_index=0

for i in range(1,len(prices)):
    if prices[i] < low_price:
        low_price=prices[i]
        low_price_index=i
    profit=prices[i]-low_price
    if profit>max_profit:
        if i >low_price_index:
            max_profit=profit
            max_profit_index=i
    
print(f"when we buy stock at Day : {low_price_index+1} at Rs:{low_price} and sell at Day:{max_profit_index+1}for {prices[max_profit_index]} we will get {max_profit}")
```
